```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import poisson
```

```python
p = 0.0003
n = 20000
lambda_ = p * n
k = np.arange(0, 100)
```

```python
pmf = poisson.pmf(k, lambda_)
cdf = poisson.cdf(k, lambda_)
```

```python
plt.bar(k, pmf)
```

```python
plt.step(k, cdf)
```
